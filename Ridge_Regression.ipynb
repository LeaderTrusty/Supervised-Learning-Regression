## Calculation of Ridge Regression Coefficient $W$

**Python Method**

**Challenge:** Refer to the formula to complete the calculation function with Python.

**Hint:** Use np.eye() to generate the identity matrix. Note that the formula requires matrix multiplication.

```python
"""Function to calculate the ridge regression coefficient
"""

import numpy as np

def ridge_regression(X, Y, alpha):
    """
    Parameters:
    X -- Matrix of independent variables
    Y -- Matrix of dependent variables
    alpha -- lambda

    Returns:
    W -- the coefficient of the ridge regression
    """

    # Calculate (X^T * X + alpha * I)
    n_features = X.shape[1]
    A = X.T @ X + alpha * np.eye(n_features)

    # Calculate W = A^-1 * X^T * Y
    W = np.linalg.inv(A) @ X.T @ Y

    return W
```

```python
np.random.seed(10) # Set the random seed

X = np.matrix(np.random.randint(5, size=(10, 10)))
Y = np.matrix(np.random.randint(10, size=(10,1 )))
alpha = 0.5
```

```python
ridge_regression(X, Y, alpha).T
```

**scikit-learn Method**

**Challenge:** Use scikit-learn to calculate the ridge regression coefficient.

**Hint:** Please add the fit_intercept=False parameter to the ridge regression model to eliminate the intercept term.

```python
"""Use scikit-learn to calculate the coefficient of the ridge regression
"""
import numpy as np
from sklearn.linear_model import Ridge

def ridge_model(X, Y, alpha):
    """
    Parameters:
    X -- Matrix of independent variables
    Y -- Matrix of dependent variables
    alpha -- lambda

    Returns:
    W -- the coefficient of the ridge regression
    """
    # This ensures your raw data is converted to the format sklearn needs
    X = np.array(X)
    Y = np.array(Y)
        # Initialize the model with fit_intercept=False
    model = Ridge(alpha=alpha, fit_intercept=False)

    # Fit the model to the data
    model.fit(X, Y)

    # Store the coefficients
    W = model.coef_

    return W
```

```python
ridge_model(X, Y, alpha)
```
